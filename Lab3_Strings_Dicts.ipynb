## Завдання 1. Варіант 1
### 1.1 Підрахунок голосних літер у рядку
Функція `count_vowels(s)` повертає кількість голосних літер у рядку. Враховуються голосні **латиниці** (a, e, i, o, u) та **кирилиці** (а, е, є, и, і, ї, о, у, ю, я) незалежно від регістру.

```python
VOWELS = set("aeiou" + "аеєиіїоуюя")


def count_vowels(s):
    if not isinstance(s, str):
        raise TypeError("Аргумент має бути рядком")
    return sum(1 for ch in s.lower() if ch in VOWELS)
```

```python
assert count_vowels("hello") == 2
assert count_vowels("AEIOU") == 5
assert count_vowels("Python") == 1
assert count_vowels("") == 0
assert count_vowels("Привіт, світ!") == 3
assert count_vowels("bcd fgh") == 0

print("Усі тести count_vowels пройдено")
```

### 1.2 Створення словника
Функція `create_dict(n)` приймає число *n* і повертає словник із *n* елементів виду `{'i': i}`, де *i* ∈ [1, n]. Ключ — рядок, значення — число.

```python
def create_dict(n):
    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("n має бути цілим числом")
    if n < 0:
        raise ValueError("n не може бути від'ємним")
    return {str(i): i for i in range(1, n + 1)}
```

```python
dict1 = create_dict(10)
dict1
```

```python
assert create_dict(3) == {'1': 1, '2': 2, '3': 3}
assert create_dict(1) == {'1': 1}
assert create_dict(0) == {}
assert len(create_dict(10)) == 10
assert create_dict(10)['10'] == 10

print("Усі тести create_dict пройдено")
```

```python
for bad in [-5, "5", 2.5, True]:
    try:
        create_dict(bad)
        print(f"{bad!r}: помилку НЕ виявлено")
    except (TypeError, ValueError) as e:
        print(f"{bad!r}: {type(e).__name__} — {e}")
```

## Завдання 2. Робота з текстом
### 2.1 Частота слів у тексті
Для тексту з файлу `text.txt` підраховується, скільки разів зустрічається кожне слово, і формується словник `<слово>: <кількість>`.

Попередня обробка тексту:
- переведення в нижній регістр;
- видалення присвійних відмінків (`'s`, `’s`);
- видалення крапок, ком, `\n`, усіх видів дужок та інших розділових знаків.

```python
import re

PUNCTUATION = r"[.,:;!?\"“”«»()\[\]{}<>/]"


def clean_text(text):
    text = text.lower()
    text = re.sub(r"['’]s\b", "", text)
    text = text.replace("\n", " ")
    text = re.sub(PUNCTUATION, " ", text)
    return text


def count_words(text):
    counts = {}
    for word in clean_text(text).split():
        counts[word] = counts.get(word, 0) + 1
    return counts
```

```python
word_counts = count_words(text)

print(f"Унікальних слів: {len(word_counts)}")
print(f"Усього слів: {sum(word_counts.values())}\n")

for word, count in list(word_counts.items())[:20]:
    print(f"{word} : {count}")
```

```python
assert count_words("Hello, hello world.") == {"hello": 2, "world": 1}
assert count_words("Python's (tutorial)\nPython") == {"python": 2, "tutorial": 1}
assert count_words("A [b] {c} <a>") == {"a": 2, "b": 1, "c": 1}
assert count_words("") == {}
assert count_words("Python’s guide, Python's docs") == {"python": 2, "guide": 1, "docs": 1}

# Перевірки на реальному тексті
assert all(word == word.lower() for word in word_counts)
assert not any(ch in word for word in word_counts for ch in ".,()[]{}\n")

print("Усі тести count_words пройдено")
```

### 2.2 Найуживаніше слово
Функція `most_common_word(d)` повертає пару (слово, кількість) з найбільшою кількістю входжень.

```python
def most_common_word(d):
    if not d:
        raise ValueError("Словник порожній")
    word = max(d, key=d.get)
    return word, d[word]


word, count = most_common_word(word_counts)
print(f"{word} : {count}")
```

### 2.3 Сортування словника за ключами (за зростанням)
### 2.4 Сортування словника за значеннями (за зростанням)

```python
def sort_by_keys(d):
    return dict(sorted(d.items()))


def sort_by_values(d):
    return dict(sorted(d.items(), key=lambda item: (item[1], item[0])))


sorted_keys = sort_by_keys(word_counts)
sorted_values = sort_by_values(word_counts)

print("За ключами (перші 10):")
for word, count in list(sorted_keys.items())[:10]:
    print(f"  {word} : {count}")

print("\nЗа значеннями (останні 10 — найчастіші):")
for word, count in list(sorted_values.items())[-10:]:
    print(f"  {word} : {count}")
```

```python
sample = {"b": 2, "a": 3, "c": 1}

# 2.2
assert most_common_word(sample) == ("a", 3)
assert most_common_word({"x": 5}) == ("x", 5)
assert most_common_word(word_counts)[1] == max(word_counts.values())

# 2.3
assert list(sort_by_keys(sample)) == ["a", "b", "c"]
assert list(sort_by_keys(word_counts)) == sorted(word_counts)

# 2.4
assert list(sort_by_values(sample).values()) == [1, 2, 3]
assert list(sort_by_values({"b": 1, "a": 1, "c": 0})) == ["c", "a", "b"]
values = list(sort_by_values(word_counts).values())
assert values == sorted(values)

# Вихідний словник не змінився
assert list(sample) == ["b", "a", "c"]

print("Усі тести 2.2–2.4 пройдено")
```

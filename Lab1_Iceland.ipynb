## Переведення чисел з десяткової системи числення у двійкову

Функція `task_7()` зчитує число (ціле або з рухомою комою) і повертає його двійкове представлення у форматі **знак|модуль**, де знак `0` означає додатне число, а `1` — від'ємне.

**Алгоритм:**
- *ціла частина* — послідовне ділення на 2 із записом остач у зворотному порядку;
- *дробова частина* — послідовне множення на 2 із записом цілих частин (не більше 10 знаків, бо деякі дроби у двійковій системі нескінченні).

Реалізовано контроль введення: порожній рядок, нечислові значення, нескінченності.

```python
import math


def int_to_bin(n):
    """Переводить ціле невід'ємне число у двійковий рядок діленням на 2."""
    if n == 0:
        return "0"
    bits = []
    while n > 0:
        bits.append(str(n % 2))
        n //= 2
    return "".join(reversed(bits))


def frac_to_bin(f, max_bits=10):
    """Переводить дробову частину (0 <= f < 1) у двійковий рядок множенням на 2."""
    bits = []
    while f > 0 and len(bits) < max_bits:
        f *= 2
        bit = int(f)
        bits.append(str(bit))
        f -= bit
    return "".join(bits)


def task_7():
    s = input().strip()

    # Контроль введення
    if not s:
        return "Помилка. Не введено жодного значення"

    try:
        value = int(s)
        is_float = False
    except ValueError:
        try:
            value = float(s)
            is_float = True
        except ValueError:
            return "Помилка. Введене значення не є числом"

    if is_float and not math.isfinite(value):
        return "Помилка. Число має бути скінченним"

    # Знак і модуль
    sign = 1 if value < 0 else 0
    value = abs(value)

    int_part = int(value)
    frac_part = value - int_part

    result = f"{sign}|{int_to_bin(int_part)}"

    if is_float:
        frac_bits = frac_to_bin(frac_part).ljust(3, "0")
        result += "." + frac_bits

    return f"{s} у двійковій системі буде {result}"
```

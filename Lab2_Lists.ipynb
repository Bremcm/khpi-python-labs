### Допоміжні функції введення
Функції зчитують дані з клавіатури та перевіряють їх коректність. Кожна повертає пару `(значення, помилка)`: якщо помилки немає, друге значення — `None`.

```python
def read_int(prompt, min_value=0):
    """Зчитує одне ціле число не менше min_value."""
    s = input(prompt).strip()
    try:
        value = int(s)
    except ValueError:
        return None, "Помилка. Потрібно ввести ціле число"
    if value < min_value:
        return None, f"Помилка. Значення має бути не менше {min_value}"
    return value, None


def read_int_list(prompt, n=None):
    """Зчитує список цілих чисел через пробіл. Якщо задано n — перевіряє кількість."""
    items = input(prompt).split()
    if n is not None and len(items) != n:
        return None, f"Помилка. Очікувалось {n} елементів, введено {len(items)}"
    try:
        numbers = [int(x) for x in items]
    except ValueError:
        return None, "Помилка. У списку присутнє нечислове значення"
    return numbers, None
```

## Завдання 1
Введіть список цілих чисел із *n* елементів. Отримайте новий список, замінивши значення **п'ятого** елемента середнім арифметичним початкового списку. *n* вводиться з клавіатури.

```python
def task_1():
    n, err = read_int("Введіть кількість елементів n: ", min_value=5)
    if err:
        return err

    numbers, err = read_int_list(f"Введіть {n} цілих чисел через пробіл: ", n)
    if err:
        return err

    mean = sum(numbers) / n
    result = numbers.copy()
    result[4] = mean
    return result
```

## Завдання 2
Введіть два списки, що складаються з *n* і *m* цілих чисел відповідно. Сформуйте **третій список** зі значень обох списків, відсортованих у порядку **спадання**. *n* і *m* також вводяться з клавіатури.

```python
def task_2():
    n, err = read_int("Введіть кількість елементів першого списку n: ")
    if err:
        return err

    list1, err = read_int_list(f"Введіть {n} цілих чисел першого списку через пробіл: ", n)
    if err:
        return err

    m, err = read_int("Введіть кількість елементів другого списку m: ")
    if err:
        return err

    list2, err = read_int_list(f"Введіть {m} цілих чисел другого списку через пробіл: ", m)
    if err:
        return err

    result = sorted(list1 + list2, reverse=True)
    return result
```

## Завдання 3
Вам надано **непорожній** список цілих чисел. Поверніть список, що складається лише з **неунікальних** елементів початкового списку (унікальні — ті, що зустрічаються лише один раз — видаляються). Порядок елементів не змінюється.

```python
from collections import Counter


def task_3():
    numbers, err = read_int_list("Введіть цілі числа через пробіл: ")
    if err:
        return err

    if not numbers:
        return "Помилка. Список не може бути порожнім"

    counts = Counter(numbers)
    result = [x for x in numbers if counts[x] > 1]
    return result
```
